```python
import sys 
import os
import platform

# Remplace les valeurs ci-dessous par TES informations
lastName = "PLANE"  
firstName = "Chloé"         
operatingSystem = "Windows11"        
duration = 4                    

# afficher les infos saisies manuellement
print("Name:", firstName, lastName)
print("Operating System:", operatingSystem)
print("Duration:", duration, "hours")

# afficher les infos système détectées automatiquement
print("platform:", platform.platform())
print("system:",   platform.system())
print("machine:",  platform.machine())
print("processor:", platform.processor())
print("node:",     platform.node())
print("  Mac :",  platform.mac_ver())
print("  Linux:",  platform.libc_ver())
print("  Windows:",  platform.win32_ver())
print("python:",  platform.python_version())

# infos sur le kernel Python
print("  kernel: ",  sys.executable)
print("  python", sys.version)

# dossier de travail actuel
print("cwd:",     os.getcwd())
```
